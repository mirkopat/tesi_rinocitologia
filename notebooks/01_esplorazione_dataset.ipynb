```python
import json
import os
from pathlib import Path

# Percorso del dataset
data_dir = Path('../data/NMCD.coco')

# Carica le annotazioni (dovresti avere 3 file: train, valid, test)
# Controlla quale file esiste
annotations_files = list(data_dir.glob('**/_annotations.coco.json'))
print(f"File di annotazioni trovati: {[f.name for f in annotations_files]}")

# Carica il primo file
if annotations_files:
    with open(annotations_files[0], 'r') as f:
        coco_data = json.load(f)
    
    print(f"Immagini: {len(coco_data['images'])}")
    print(f"Annotazioni: {len(coco_data['annotations'])}")
    print(f"Categorie: {len(coco_data['categories'])}")
    
    # Mostra le categorie
    for cat in coco_data['categories']:
        print(f"  {cat['id']}: {cat['name']}")
```

```python
from collections import Counter

for split in ['train', 'valid', 'test']:
    ann_file = data_dir / split / '_annotations.coco.json'
    with open(ann_file, 'r') as f:
        data = json.load(f)
    
    class_counts = Counter()
    for ann in data['annotations']:
        class_counts[ann['category_id']] += 1
    
    print(f"\n{split}:")
    for cat_id, count in class_counts.most_common():
        cat_name = next(cat['name'] for cat in data['categories'] if cat['id'] == cat_id)
        print(f"  {cat_name}: {count}")
```
